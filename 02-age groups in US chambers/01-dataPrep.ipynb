# Senate and House of Reps

1. Open pages of wikipedia articles:
    - https://en.wikipedia.org/wiki/List_of_current_United_States_representatives
    - https://en.wikipedia.org/wiki/List_of_current_United_States_senators
2. View html-code, select tables, copy, and paste into files.
3. Clean up and export as csv files.

## read, convert, and clean the data

```python
def readWikipediaTables(filename):
    """Read the tables that were copied from the respective wikipedia pages."""
    # read table, clean up columns and column names, split column "Born" into date and age
    data1 = pd.read_html(filename)[0]
    data1 = data1.drop(columns=["Party"]) # useless column, it contained party as color
    data1 = data1.rename(columns={"Born[4]":"Born","Party.1":"Party"})
    data1[['BirthMonth', 'BirthDay','BirthYear','age']] = data1["Born"].str.split(" |, ",expand=True)
    data1.age = 2026 - pd.to_numeric(data1.BirthYear)
    data1.Party = data1.Party.replace('Independent[o]','Independent') # correcting mistake in table

    
    return data1

def readCleanWrite_WikiHTMLtoCSV():
    
    # data copied from https://en.wikipedia.org/wiki/List_of_current_United_States_representatives on Oct 7 2026
    # titles read `List of representatives as of September 2, 2026`
    filenameReps = "data/List_of_current_United_States_representatives.html" 
    reps = readWikipediaTables(filenameReps)
    reps.to_csv("data/HouseOfReps.csv")

    # data copied from https://en.wikipedia.org/wiki/List_of_current_United_States_senators on Oct 7 2026
    filenameSenate = "data/List_of_current_United_States_senators.html" 
    senate = readWikipediaTables(filenameSenate)
    senate.to_csv("data/HouseOfSenate.csv")

def main():
    print("Reading HTML raw data and write to csv...")
    try:
        readCleanWrite_WikiHTMLtoCSV()
        print("Done")
    except:
        print("Error")
```

```python
main()
```
